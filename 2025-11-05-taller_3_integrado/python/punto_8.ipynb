```python
import speech_recognition as sr

r = sr.Recognizer()

def escuchar():
    with sr.Microphone() as source:
        print("🎤 Diga un comando...")
        r.adjust_for_ambient_noise(source)
        audio = r.listen(source)

    try:
        # Opción 1: reconocimiento offline con CMU Sphinx
        comando = r.recognize_sphinx(audio, language="es-ES")
        print(f"Comando detectado: {comando}")
        return comando.lower()
    except sr.UnknownValueError:
        print("No se entendió el comando.")
    except sr.RequestError:
        print("Error al usar Sphinx.")
```

```python
from pythonosc import udp_client
import pyttsx3

# Cliente OSC para enviar datos a Unity/Processing
client = udp_client.SimpleUDPClient("127.0.0.1", 5005)
voz = pyttsx3.init()

comandos = {
    "adelante": lambda: client.send_message("/movimiento", "adelante"),
    "atrás": lambda: client.send_message("/movimiento", "atras"),
    "izquierda": lambda: client.send_message("/movimiento", "izquierda"),
    "derecha": lambda: client.send_message("/movimiento", "derecha"),
    "saludar": lambda: voz.say("Hola, ¿cómo estás?")
}

def ejecutar_comando(cmd):
    for palabra, accion in comandos.items():
        if palabra in cmd:
            print(f"→ Ejecutando acción: {palabra}")
            accion()
            voz.runAndWait()
            break
```

```python
def main():
    while True:
        comando = escuchar()
        if comando:
            ejecutar_comando(comando)
            if "salir" in comando:
                voz.say("Saliendo del programa.")
                voz.runAndWait()
                break

if __name__ == "__main__":
    main()
```
